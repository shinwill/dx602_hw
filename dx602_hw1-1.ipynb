```python
# Problem 1 variables and assignments

my_int = 26
my_float = 10.26
my_string = 'pumpkin'

print('my favorite integer is', my_int)
print('my favorite float is', my_float)
print('my favorite string is', my_string)
```

```python
# Problem 2 arithmetic operations

x = 26 + 52
y = x * 5 

print('x =', x)
print('y =', y)
```

```python
# Problem 3 multiplication and division of floating point numbers

width = 2.5
height = 5.2 

area = width * height
print('area =', area)

ratio = width / height
print('ratio =', ratio)
```

```python
# Problem 4 division that results a floating-point number 

cookies = 48
players = 13

cookies_per_player = cookies / players

# print(cookies_per_player)
print('Each player gets', cookies_per_player, 'cookies.')
```

```python
# Problem 5 arithmetic with integer division and remainder

cookies = 48
players = 13

whole_cookies_per_player = cookies // players
print('Each player gets', whole_cookies_per_player, 'cookies.')

cookies_left_over = cookies % players 
print('That leaves', cookies_left_over, 'cookies for the coach!')
```

```python
# Problem 7 extracting substrings using string slicing 

color = 'darkblue'

first_half = color[0:4]
last_half = color[-4:]

print('first half:', first_half)
print('last half:', last_half)
```

```python
# Problem 11 using built in functions on lists 

values = [6,1,7,3,5,8,2,3,6,1]

length = len(values)
total = sum(values)
mean = total / length 

print('The list contains', length, 'elements, the sum is', total, 'and the mean is', mean)
```

```python
# Problem 12 assessing elements by position 

values = [6,1,7,3,5,8,2,3,6,1]

first = values[0]
middle = values[4], values[5]
last = values[-1]

print('The first value is', first)
print('The middle values are', middle)
print('The last value is', last)
```

```python
# Problem 13

values = [6,1,7,3,5,8,2,3,6,1]

first_three = values[0:3]
last_three = values[-3:]

#print(last_three)

combined = first_three + last_three

print('The first 3 elements and the last 3 elements combined are :', combined)

length = len(combined)
total = sum(combined)

mean_combined = total/length
print('And the mean of the combined list is', mean_combined)
```
